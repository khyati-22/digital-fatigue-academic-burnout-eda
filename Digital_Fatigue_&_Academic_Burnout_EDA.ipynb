## 1. Project Introduction

### 1.1 Background

The increasing use of digital devices has become an important part of students' academic and personal lives. Students regularly use smartphones, laptops, and other digital devices for online learning, assignments, social media, entertainment, and other activities.

While digital technology provides several academic and social benefits, prolonged screen exposure may also be associated with symptoms such as eye strain, headaches, difficulty concentrating, mental exhaustion, and reduced productivity. These experiences may contribute to digital fatigue and may be related to students' academic well-being and burnout.

Academic burnout can involve emotional and mental exhaustion, stress related to academic workload, reduced motivation, and feelings of disconnection from studies.

This project uses primary survey data collected through a structured questionnaire to explore patterns of digital usage, digital fatigue, academic well-being, and academic burnout among students.

# 4. Data Cleaning & Preprocessing

Data cleaning is an important step before performing Exploratory Data Analysis (EDA). It ensures that the dataset is consistent, accurate, and suitable for further analysis.

The following data-cleaning steps are performed:

- Creating a working copy of the original dataset
- Checking for missing values
- Checking for duplicate responses
- Removing unnecessary whitespace from column names and text responses
- Standardizing inconsistent categorical responses
- Checking unique values in categorical variables
- Verifying the final cleaned dataset

```python
df_clean = df.copy()

print("Clean dataset created successfully!")
print("Original dataset shape:", df.shape)
print("Working dataset shape:", df_clean.shape)
```

### 4.1 Standardizing Column Names

The questionnaire responses contain some unnecessary spaces in column names. These spaces can create difficulties when accessing variables during analysis.

Therefore, leading and trailing spaces are removed from all column names.

```python
# Remove leading and trailing spaces from column names

df_clean.columns = df_clean.columns.str.strip()

print("Column names standardized successfully!")
```

### 4.2 Checking Missing Values

Missing values can affect statistical analysis and visualization. Therefore, the dataset is checked for missing values before proceeding with further analysis.

```python
# Check missing values in each column

missing_values = df_clean.isnull().sum()

missing_values[missing_values > 0]
```

```python
# Calculate total missing values

total_missing = df_clean.isnull().sum().sum()

print("Total missing values in the dataset:", total_missing)
```

```python
# Display missing values along with their percentage

missing_summary = pd.DataFrame({
    'Missing Values': df_clean.isnull().sum(),
    'Missing Percentage': (df_clean.isnull().sum() / len(df_clean)) * 100
})

missing_summary = missing_summary[missing_summary['Missing Values'] > 0]

missing_summary.sort_values(by='Missing Values', ascending=False)
```

### Interpretation of Missing Values

A total of 40 missing values were identified in the dataset. All 40 missing values belong to the open-ended question asking respondents to provide suggestions for reducing digital fatigue and academic burnout.

Since this was an open-ended question, respondents were not required to provide a response. Therefore, these missing values are treated as unanswered responses rather than errors.

The corresponding respondents will not be removed from the dataset because their responses to the other questionnaire variables are still valid and useful for analysis.

No imputation will be performed for this open-ended variable.

### 4.3 Checking Duplicate Responses

Duplicate responses can affect the accuracy of the analysis because the same respondent may be counted more than once. Therefore, the dataset is checked for completely duplicated records before proceeding with further analysis.

```python
# Check the number of completely duplicated rows

duplicate_count = df_clean.duplicated().sum()

print("Number of duplicate responses:", duplicate_count)
```

```python
# Display duplicate rows, if any

duplicate_rows = df_clean[df_clean.duplicated(keep=False)]

duplicate_rows
```

No completely duplicated responses were identified in the dataset. Therefore, no observations were removed due to duplication.

### 4.4 Cleaning Text Responses

Text-based questionnaire responses may contain leading or trailing spaces. These spaces can result in identical responses being treated as different categories.

Therefore, whitespace is removed from text responses.

```python
# Remove leading and trailing spaces from text values

for column in df_clean.select_dtypes(include='object').columns:
    df_clean[column] = df_clean[column].str.strip()

print("Extra spaces removed from text responses.")
```

```python
# Standardize capitalization for Likert-scale responses

likert_columns = [
    'I experience eye strain after screen use.',
    'I frequently experience headaches due to screen exposure.',
    'These headaches affect my productivity or daily activities.',
    'I find it difficult to focus on offline tasks after prolonged screen exposure.',
    'I feel emotionally drained due to my academic workload.',
    'I feel disconnected from my studies.',
    'I feel stressed when thinking about assignments and deadlines.',
    'I find it difficult to stay motivated towards my studies.',
    'How would you rate your sleep quality?',
    'How often do you use digital devices right before going to sleep?',
    'How would you rate your overall stress level?',
    'How effective do you find these methods in reducing fatigue?',
    'How often do you take breaks while using digital devices?',
    'Are you aware of the negative effects of prolonged screen time on mental and physical health?',
    'How often do you feel mentally exhausted due to your studies?',
    'How would you rate your motivation towards academic work?'
]

for column in likert_columns:
    if column in df_clean.columns:
        df_clean[column] = df_clean[column].apply(
            lambda x: x.strip().title() if isinstance(x, str) else x
        )

print("Likert-scale responses standardized successfully.")
```

```python
# Check unique responses after standardization

for column in likert_columns:
    if column in df_clean.columns:
        print("\n" + column)
        print(df_clean[column].unique())
```

### 4.5 Checking Categorical Responses

After standardizing the text responses, the unique categories of important questionnaire variables are examined to identify any remaining inconsistencies or unexpected values.

```python
# Display unique values for selected categorical variables

categorical_columns = [
    'Course',
    'Mode of Study',
    'Which device do you primarily use for your daily activities?',
    'On average, how many hours do you spend on digital devices daily?',
    'What is the primary purpose of your screen usage?',
    'How many hours per day do you spend on attending online activities?'
]

for column in categorical_columns:
    if column in df_clean.columns:
        print("\n" + column)
        print(df_clean[column].unique())
```

### 5.3 Frequency Distribution of Categorical Variables

Frequency distribution shows how many respondents selected each category for a particular question.

This helps identify the most common responses and understand the overall response pattern of the participants.

```python
# Function to display frequency and percentage distribution

def frequency_table(column):
    frequency = df_clean[column].value_counts(dropna=False)
    percentage = df_clean[column].value_counts(normalize=True, dropna=False) * 100

    summary = pd.DataFrame({
        'Frequency': frequency,
        'Percentage': percentage.round(2)
    })

    return summary
```

# 8. Digital Fatigue Analysis

This section focuses on analysing the physical and cognitive symptoms associated with digital fatigue among students.

Four questionnaire variables are used as the main indicators of digital fatigue:

1. Eye strain after screen use
2. Headaches due to screen exposure
3. Headaches affecting productivity or daily activities
4. Difficulty focusing on offline tasks after prolonged screen exposure

The analysis includes the creation of a Digital Fatigue Score, classification of respondents into fatigue categories, distribution analysis, and examination of the relationship between digital fatigue and screen time, sleep quality, break frequency, and academic burnout.

### Digital Fatigue Indicators

The following four Likert-scale questions are selected as indicators of digital fatigue:

- Eye strain after screen use
- Headaches due to screen exposure
- Headaches affecting productivity or daily activities
- Difficulty focusing on offline tasks after prolonged screen exposure

Each variable uses a five-point Likert scale ranging from Strongly Disagree to Strongly Agree.

```python
fatigue_columns = [
    'I experience eye strain after screen use.',
    'I frequently experience headaches due to screen exposure.',
    'These headaches affect my productivity or daily activities.',
    'I find it difficult to focus on offline tasks after prolonged screen exposure.'
]

# Display the responses
df_clean[fatigue_columns].head()
```

### Response Distribution of Digital Fatigue Indicators

Before calculating the overall Digital Fatigue Score, the response distribution for each indicator is examined.

This helps identify how frequently respondents experience individual symptoms of digital fatigue.

```python
for column in fatigue_columns:
    print("\n" + column)
    print(df[column].value_counts())
```

### Likert Scale Conversion

To calculate an overall Digital Fatigue Score, the responses are converted into numerical values.

The scoring system is:

- Strongly Disagree = 1
- Disagree = 2
- Neutral = 3
- Agree = 4
- Strongly Agree = 5

A higher score indicates a higher level of reported digital fatigue.

```python
likert_mapping = {
    'strongly disagree': 1,
    'disagree': 2,
    'neutral': 3,
    'agree': 4,
    'strongly agree': 5
}

# Convert responses into numerical values
fatigue_scores = pd.DataFrame()

for column in fatigue_columns:
    fatigue_scores[column] = (
        df_clean[column]
        .astype(str)
        .str.strip()
        .str.lower()
        .map(likert_mapping)
    )

fatigue_scores.head()
```

### Digital Fatigue Score

The Digital Fatigue Score is calculated by adding the scores of the four digital fatigue indicators.

Digital Fatigue Score =

Eye Strain + Headache + Productivity Impact + Concentration Difficulty

Since each indicator ranges from 1 to 5, the total score ranges from 4 to 20.

A higher score represents greater reported digital fatigue.

```python
df_clean['Digital Fatigue Score'] = fatigue_scores.sum(axis=1)

# Display score
df_clean[['Digital Fatigue Score']].head(10)
```

Descriptive statistics are used to understand the overall distribution of Digital Fatigue Scores among respondents.

### Digital Fatigue Categories

The possible Digital Fatigue Score ranges from 4 to 20.

For transparent interpretation, respondents are classified into three categories:

- Low Digital Fatigue: 4–9
- Moderate Digital Fatigue: 10–15
- High Digital Fatigue: 16–20

These categories are based on the possible score range and are used consistently throughout the analysis.

```python
def classify_fatigue(score):
    if score <= 9:
        return 'Low'
    elif score <= 15:
        return 'Moderate'
    else:
        return 'High'

df_clean['Fatigue Category'] = df_clean['Digital Fatigue Score'].apply(classify_fatigue)

df_clean[['Digital Fatigue Score', 'Fatigue Category']].head(10)
```

### Likert Scale Encoding

For agreement-based questions:

1 = Strongly Disagree  
2 = Disagree  
3 = Neutral  
4 = Agree  
5 = Strongly Agree

For mental exhaustion:

1 = Never  
2 = Rarely  
3 = Sometimes  
4 = Often  
5 = Always

```python
likert_columns = [
    'Emotional_Exhaustion',
    'Disconnection',
    'Academic_Stress',
    'Motivation_Difficulty'
]

for col in likert_columns:
    burnout_wellbeing[col] = (
        burnout_wellbeing[col]
        .astype(str)
        .str.strip()
        .str.title()
    )

agreement_mapping = {
    'Strongly Disagree': 1,
    'Disagree': 2,
    'Neutral': 3,
    'Agree': 4,
    'Strongly Agree': 5
}

for col in likert_columns:
    burnout_wellbeing[col + '_Score'] = (
        burnout_wellbeing[col].map(agreement_mapping)
    )

print("Likert encoding completed.")
```
